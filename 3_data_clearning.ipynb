```python
import pandas as pd

df = pd.read_csv("student_focus_dropoff.csv")
```

```python
df.isnull().sum()

```

```python
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include="object").columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
df[df["avg_watch_pct"] > 100]
df[df["avg_watch_pct"] < 0]

```

```python
df[df["login_days_14d"] > 14]
```

```python
df[df["last_active_days_ago"] < 0]
```

```python
df["avg_watch_pct"] = df["avg_watch_pct"].clip(0, 100)
df["login_days_14d"] = df["login_days_14d"].clip(0, 14)
```
