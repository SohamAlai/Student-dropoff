```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_students = 5000

data = pd.DataFrame({
    "videos_watched_14d": np.random.poisson(lam=8, size=n_students),
    "avg_watch_pct": np.clip(np.random.normal(65, 20, n_students), 10, 100),
    "avg_session_minutes": np.clip(np.random.normal(35, 15, n_students), 5, 120),
    "login_days_14d": np.random.randint(1, 15, n_students),
    "assignments_submitted": np.random.poisson(lam=3, size=n_students),
    "avg_submission_delay": np.clip(np.random.exponential(scale=2, size=n_students), 0, 10),
    "quiz_score_avg": np.clip(np.random.normal(70, 15, n_students), 30, 100),
    "quiz_score_trend": np.random.choice(
        ["improving", "stable", "declining"],
        size=n_students,
        p=[0.35, 0.40, 0.25]
    ),
    "forum_activity": np.random.poisson(lam=2, size=n_students),
    "last_active_days_ago": np.random.randint(0, 14, n_students)
})

# Drop-off logic (probabilistic)
drop_prob = (
    0.35 * (data["login_days_14d"] < 4).astype(int) +
    0.25 * (data["avg_watch_pct"] < 40).astype(int) +
    0.20 * (data["quiz_score_trend"] == "declining").astype(int) +
    0.20 * (data["last_active_days_ago"] > 5).astype(int)
)

data["drop_off"] = np.random.binomial(1, np.clip(drop_prob, 0, 1))

data.head()
```

```python
data.to_csv("student_focus_dropoff.csv", index=False)

```
