```python
import pandas as pd

df = pd.read_csv("student_focus_dropoff_cleaned.csv")
```

```python
df["videos_per_login"] = df["videos_watched_14d"] / (df["login_days_14d"] + 1)
```

```python
df["assignment_delay_flag"] = (df["avg_submission_delay"] > 3).astype(int)
```

```python
df["low_session_quality"] = (
    (df["avg_session_minutes"] < 20) & (df["avg_watch_pct"] < 40)
).astype(int)
```

```python

df["inactive_risk"] = (df["last_active_days_ago"] > 5).astype(int)
```

```python
df["low_quiz_score"] = (df["quiz_score_avg"] < 50).astype(int)
```

```python
X = df.drop("drop_off", axis=1)
y = df["drop_off"]
```

```python
df.to_csv("student_focus_dropoff_engineered.csv", index=False)

```
